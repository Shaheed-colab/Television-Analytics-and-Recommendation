```python
def fast_charging(x):
    if x <= 30:
        return 'slow'
    elif x >=30 and x <= 60:
        return 'medium'
    else:
        return 'high'
        
charging = df1['fast_charging'].apply(fast_charging)
model_pos = df1.columns.get_loc('ram')
df1.insert(model_pos, 'charging_category', charging)
```
