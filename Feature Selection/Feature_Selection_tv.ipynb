```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('television_outlier_detection.csv')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
df.head()
```

```python
df['brand_score'].value_counts()
```

```python
def brand_category(x):
    if x >=1 and x <= 3:
        return 'small'
    elif x >= 3 and x <= 6:
        return 'medium'
    else:
        return 'big'
```

```python
df['brand_category'] = df['brand_score'].apply(brand_category)
```

```python
df.head()
```

```python
df1 = df.drop(columns=['Unnamed: 0','brand_score'])
```

```python
df1.head()
```

```python
df1.info()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
data_label_encoded = df1.copy()

cat_col = df1.select_dtypes(include = ['object']).columns

for col in cat_col:
    oe = OrdinalEncoder()
    data_label_encoded[col] = oe.fit_transform(data_label_encoded[[col]]) 
    print(oe.categories_)
    
X_label = data_label_encoded.drop(columns = ['Price'])
y_label = data_label_encoded['Price']
```

```python
X_label.head()
```

```python
y_label.head()
```

# Technique - 1 Correlation Analysis

```python
sns.heatmap(data_label_encoded.corr())
```

```python
f1_df1 = data_label_encoded.corr()['Price'].iloc[1:].to_frame().reset_index().rename(columns = {'index':'features', 'Price':'coef_corr'})
f1_df1
```

# Technique - 2 Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(X_label, y_label)

f2_df2 = pd.DataFrame({
    'features' : X_label.columns,
    'rf_importance' : rf.feature_importances_
}).sort_values('rf_importance', ascending=False)
```

```python
f2_df2
```

# Technique - 3 Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gdr = GradientBoostingRegressor(n_estimators = 100, random_state = 42)
gdr.fit(X_label, y_label)

f3_df3 = pd.DataFrame({
    'features' : X_label.columns,
    'gdr_importance' : gdr.feature_importances_
}).sort_values('gdr_importance', ascending=False)

f3_df3
```

# Technique - 4 Permutation Importance

```python
from sklearn.inspection import permutation_importance
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_label, y_label, test_size=0.2, random_state=42)

rf_label = RandomForestRegressor(n_estimators = 100, random_state = 42)
rf_label.fit(X_train, y_train)

pm = permutation_importance(rf_label, X_test, y_test, n_repeats = 30, random_state = 42)

f4_df4 = pd.DataFrame({
    'features' : X_label.columns,
    'pm_importance' : pm.importances.mean(axis=1)
}).sort_values('pm_importance', ascending=False)

f4_df4
```

# Technique - 5 Lasso Importance

```python
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
x_label_scale = sc.fit_transform(X_label)

lasso = Lasso(alpha = 0.01, random_state=42)
lasso.fit(x_label_scale, y_label)

f5_df5 = pd.DataFrame({
    'features' : X_label.columns,
    'lasso_importance' : lasso.coef_
}).sort_values('lasso_importance', ascending=False)

f5_df5
```

# Technique - 6 RFE Importance

```python
from sklearn.feature_selection import RFE
```

```python
estimator = RandomForestRegressor()
rfe_label = RFE(estimator, n_features_to_select = X_label.shape[1], step = 1)
rfe_label.fit(X_label, y_label)

select_rfe_features = X_label.columns[rfe_label.support_]

selected_estimator = rfe_label.estimator_.feature_importances_

f6_df6 = pd.DataFrame({
    'features' : select_rfe_features,
    'rfe_importance' : selected_estimator
}).sort_values('rfe_importance', ascending=False)

f6_df6
```

# Technique - 7 LinearRegression Importance

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_label, y_label)

f7_df7 = pd.DataFrame({
    'features' : X_label.columns,
    'lr_importance' : lr.coef_
}).sort_values('lr_importance', ascending=False)

f7_df7
```

```python
final_df = f1_df1.merge(f2_df2, on = 'features').merge(f3_df3, on = 'features').merge(f4_df4, on = 'features').merge(f5_df5, on = 'features').merge(f6_df6, on = 'features').merge(f7_df7, on = 'features')
```
