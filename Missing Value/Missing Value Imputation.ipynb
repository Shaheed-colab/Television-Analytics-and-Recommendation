```python
numeric_cols = ['resolution_width', 'resolution_height', 'fast_charging', 'ram', 'rom', 'Spec_Score', 'usb_port']
```

```python
from sklearn.impute import KNNImputer
```

```python
knn = KNNImputer(n_neighbors=5)
knn_impute = knn.fit_transform(df1[numeric_cols])
```

```python
imputed_df = pd.DataFrame(knn_impute,
                         columns=numeric_cols,
                         index=df1.index)
```

```python
imputed_df.isnull().sum()
```

```python
df2 = pd.concat([df1.drop(columns = numeric_cols), imputed_df], axis=1)
```
