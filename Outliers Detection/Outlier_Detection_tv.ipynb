```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('television_eda.csv')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
df.head()
```

```python
df = df.drop(columns = ['Unnamed: 0'])
```

```python
df.head()
```

# Price

```python
df['Price'].describe()
```

```python
sns.distplot(df['Price'], bins=10)
```

```python
sns.boxplot(data=df, x='Price')
```

```python
upper_limit = df['Price'].quantile(0.95)
upper_limit
```

```python
lower_limit = df['Price'].quantile(0.05)
lower_limit
```

```python
df[(df['Price'] <= 12.34) & (df['Price'] >= 9.12)]
```

```python
df['Price'] = np.where(df['Price'] >= upper_limit,
                      upper_limit,
                      np.where(df['Price'] <= lower_limit,
                              lower_limit,
                              df['Price']))
```

```python
sns.boxplot(data=df, x='Price')
```
