```python
tv_data = pd.read_csv('television_outlier_detection.csv')
rs_data = pd.read_csv('rs.csv')
```

```python
first_data = pd.read_csv('Televisions.csv')
first_data.head()
```

```python
tv_data = tv_data.drop(columns = ['Unnamed: 0'])
rs_data = rs_data.drop(columns = ['Unnamed: 0'])
```

```python
tv_data.head()
```

```python
rs_data.head()
```

```python
total_df = tv_data.merge(first_data, on='Rating', how='left')
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
total_df.head()
```

```python
total_df.columns
```

```python
df = total_df.drop(columns=['index', 'Company_amazon', 'Company_croma', 'Company_flipkart'])
df.head()
```

```python
cols = [
    'ultra_hd',
    'full_hd',
#     'Price',
#     'Rating',
    'inches',
    'display_type',
    'wifi',
    'rom',
    'hdmi_port',
    'value_score',
    'smart_features',
    'ppi',
    'resolution_width',
    'resolution_height',
    'fast_charging',
    'ram',
    'usb_port',
#     'Spec_Score',
    'size_category',
    'charging_category',
    'brand_score'
]
```

```python
df['tags'] = df[cols].astype(str).agg(' '.join, axis=1)
```

```python
df.head()
```

```python
new_df = df[['Model', 'tags', 'Price_y', 'Rating', 'Spec_Score_y']]
```

```python
new_df.head()
```

```python
final_data = new_df.merge(rs_data, on='Model', how = 'left')
```

```python
final_data.head()
```

```python
final_data['Price_y'] = final_data['Price_y'].str.replace('₹', '').str.replace(',', '')
```

```python
final_data.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
```

```python
vectors = cv.fit_transform(final_data['tags']).toarray()
```

```python
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(name):
    index = final_data[final_data['Model'] == name].index[0]

    similarity = cosine_similarity(
        vectors[index:index+1],
        vectors
    ).flatten()

    distances = sorted(
        list(enumerate(similarity)),
        reverse=True,
        key=lambda x: x[1]
    )

    for i in distances[1:6]:
        print(final_data.iloc[i[0]]['Model'])
        print(final_data.iloc[i[0]]['Price_y'])
```

```python
recommend('TCL P7K 55 inch Ultra HD 4K Smart QLED TV (55P7K)')
```
